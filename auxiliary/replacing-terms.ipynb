```python
def replace_terms_in_file(input_file_path, replacements, output_file_path=None):
    with open(input_file_path, 'r', encoding='utf-8') as file:
        content = file.read()

    for old_term, new_term in replacements.items():
        content = content.replace(old_term, new_term)

    if output_file_path is None:
        output_file_path = input_file_path

    with open(output_file_path, 'w', encoding='utf-8') as file:
        file.write(content)

    print(f"Terms replaced and file saved to {output_file_path}")

input_file_path = 'sequenceMining3.txt'
output_file_path = 'newSequenceMining3.txt'

replacements = {
    'M001': '6 -13',
    'M002': '0 -11',
    'M003': '0 -9',
    'M004': '9 -5',
    'M005': '2 -6',
    'M006': '-2 -7',
    'M007': '3 -9',
    'M008': '-4 -7',
    'M009': '-7 -11',
    'M010': '-7 -13',
    'M011': '-4 -16',
    'M012': '-13 -14',
    'M013': '-11 -5',
    'M014': '-11 6',
    'M015': '-14 9',
    'M016': '-10 16',
    'M017': '-10 14',
    'M018': '-10 9',
    'M019': '-12 12',
    'M020': '-10 -6',
    'M021': '-4 5',
    'M022': '2 6',
    'M023': '6 4',
    'M024': '12 2',
    'M025': '15 10',
    'M026': '11 15',
    'M027': '11 12',
    'M028': '6 10',
    'M029': '3 12',
    'M030': '4 16',
    'M031': '-3 9',
    ',': ' '
}
```
